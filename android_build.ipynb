## 3. Build the APK (slow first time, ~30-45 min)

Buildozer will download the Android SDK, NDK, build tools, Python-for-Android, and your dependencies. Watch the output — if it fails, the error is usually near the bottom and tells you what's missing.

**Don't close the Colab tab while this runs.** Colab will idle-disconnect if your browser tab is hidden for too long.

```python
!yes | buildozer -v android debug 2>&1 | tail -300
```

## 4. Find the built APK

```python
import glob
apks = sorted(glob.glob('bin/*.apk'))
for apk in apks:
    import os
    size_mb = os.path.getsize(apk) / (1024 * 1024)
    print(f'{apk}  ({size_mb:.1f} MB)')

if not apks:
    print('No APK found. Build probably failed — scroll up in cell 3 output to find the error.')
```

## 5. Download the APK to your Mac

```python
from google.colab import files
import glob

apks = sorted(glob.glob('bin/*.apk'))
if apks:
    files.download(apks[-1])
    print(f'Downloading {apks[-1]} to your Downloads folder...')
else:
    print('No APK to download. Check cell 3 output for build errors.')
```
